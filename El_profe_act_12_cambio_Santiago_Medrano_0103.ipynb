## Las variables
1. CP --> Costo del producto
2. CD --> Cantidad de dinero
3. Cambio
## Formulas
* $Cambio = CD - CP$
## Diagrama de flujo LucidChart
<img src="https://raw.githubusercontent.com/SantiagoMedrano128/Calcular-cambio-if-else-/refs/heads/main/DF%20if%20else%20Cambio%202.png" alt="Imagen aleatoria" width="350" height="500">

```python
# Para cuando es falso (no)
# Calculando el cambio usando if else
# Datos de entrada CD, CP
# Mostrar cambio si es falso
# Si es verdadero mostrar no alcanza
CP = float(input("Ingrese el costo del producto: "))
CD = float(input("Ingrese la cantidad de dinero: "))
if CP > CD:
    print("No alcanza")
else:
    Cambio = CD - CP
    print("El cambio es: ", Cambio)
print("Fin del programa")
print("Programa realizao por Santiago Medranno NL = 37")
```

```python
# Para cuando es verdadero (si)
# Calculando el cambio usando if else
# Datos de entrada CD, CP
# Mostrar cambio si es falso
# Si es verdadero mostrar no alcanza
CP = float(input("Ingrese el costo del producto: "))
CD = float(input("Ingrese la cantidad de dinero: "))
if CP > CD:
    print("No alcanza")
else:
    Cambio = CD - CP
    print("El cambio es: ", Cambio)
print("Fin del programa")
print("Programa realizao por Santiago Medranno NL = 37")
```
